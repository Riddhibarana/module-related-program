## Question 2

Write a python program to make a module which contain all the basic functions related to string and import that module in another file and use that fuctions with string given by user.

*Difficulty: Medium*

```python
%%writefile stringmod.py
def upper(s):
    return s.upper()

def lower(s):
    return s.lower()

def title(s):
    return s.title()

def reverse(s):
    return s[::-1]

def length(s):
    return len(s)

def count(s, ch):
    return s.count(ch)
```

```python
%%writefile main.py
from stringmod import *
s=input("enter string")
print("Uppercase:", upper(s))
print("Lowercase:", lower(s))
print("Title Case:", title(s))
print("Reverse:", reverse(s))
print("Length:", length(s))

ch = input("Enter a character to count: ")
print("Count:", count(s, ch))
```

Write a python program to make a module named cal.py which contain all the basic functions related to calculator like addition, subtraction, multiplication, and division import that module in another file and use that functions with number inputs given by user.

*Difficulty: Medium*

```python
%%writefile call.py

def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        return "Cannot divide by zero"
    return a / b
```

```python
import call

a = float(input("Enter first number: "))
b = float(input("Enter second number: "))

print("Addition:", call.add(a, b))
print("Subtraction:", call.subtract(a, b))
print("Multiplication:", call.multiply(a, b))
print("Division:", call.divide(a, b))
```

## Question 5

Write a program to create a module ‘first_word.py’, which returns the first word of any string passed. Show the working of the module, by calling the module with any suitable example.
Input: ‘This is Python Programming’
Output: ‘This’

*Difficulty: Medium*

```python
%%writefile first_word.py

def get_first_word(string):
    words = string.split()
    if words:
        return words[0]
    return ""
```

```python
import first_word

s = input("Enter a string: ")

result = first_word.get_first_word(s)

print("First word:", result)
```
